# The Geometric structure of networks

## Introduction to network science

Network science is the study of entities through their relationships with each other. Mathematically, networks are rooted in graph theory, which distills entities and relationships down to their abstract essence. Facebook networks, can be defined as the smallest number of friendships connecting two individuals. This is a useful way of measuring a users centrality in a network which can define their influence. A persons sphere of influence can be measure to detect inauthentic accounts. Such as identify users with a small circle of friends, but has a large network of friends of friends. Using network analysis, people were able to identify gerrymandering.
